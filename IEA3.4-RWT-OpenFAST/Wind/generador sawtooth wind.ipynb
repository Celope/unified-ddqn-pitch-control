```python
import numpy as np


def generar_viento_sawtooth(
    nombre_archivo="sawtooth_w.wnd", tiempo_max=5500, paso=10
):
    # 1. Crear el vector de tiempo (de 0 a 2230 segundos, cada 10 segundos)
    tiempo = np.arange(0, tiempo_max + paso, paso)

    # 2. Generar la velocidad del viento alternando entre 14.5 y 15.5
    # Si el índice es par (0, 20, 40...) da 14.5. Si es impar (10, 30, 50...) da 15.5
    velocidad_viento = np.where(
        (tiempo // paso) % 2 == 0, 7.5, 8.5
    )

    # 3. Columnas de ceros para los demás parámetros requeridos por el formato
    ceros = np.zeros_like(tiempo, dtype=float)

    # 4. Escribir el archivo con el formato original
    with open(nombre_archivo, "w") as f:
        # Cabeceras estándar IEC
        f.write("!OUTPUT FROM, IEC 61400-1 2nd EDITION.\n")
        f.write(
            "!---------------------------------------------------------------\n"
        )
        f.write("!Stats for this wind file:\n")
        f.write("!     Only Horizontal Velocity\n")
        f.write(
            "!---------------------------------------------------------------\n"
        )
        f.write(
            " ! Time\t\tWind\t\tWind\t\tVertical\tHoriz.\t\tPwr.Law\t\tLin.Vert.\tGust\n"
        )
        f.write(
            " ! Speed\tDir\t\tSpeed\t\tShear\t\tVert.Shr\tShear\t\tSpeed\n"
        )
        f.write(
            " ! (sec)\t(m/s)\t\t(Deg)\t\t(m/s)\t\t\t\t\t\t\t(m/s)\n"
        )

        # Escribir los datos línea por línea con tabulaciones limpias
        for t, v in zip(tiempo, velocidad_viento):
            f.write(
                f"{t}\t{v:.1f}\t    {ceros[0]:.3f}\t   {ceros[0]:.3f}\t    {ceros[0]:.3f}\t    {ceros[0]:.3f}\t    {ceros[0]:.3f}\t    {ceros[0]:.3f}\n"
            )

    print(f"¡Archivo '{nombre_archivo}' generado con éxito!")


# Ejecutar la función
generar_viento_sawtooth()
```
